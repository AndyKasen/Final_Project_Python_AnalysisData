3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
#ваш код здесь
def exp_work(text):
    if pd.isna(text):
        return np.nan
    
    text_str = str(text).lower()
    exp_match = re.search(r'опыт\s+работы\s+([^\.\n]+)', text_str)
    if not exp_match:
        return np.nan
    
    exp_text = exp_match.group(1)
    
    years = months = 0
    year_match = re.search(r'(\d+)\s*(?:год|года|лет)', exp_text)
    if year_match:
        years = int(year_match.group(1))
    
    month_match = re.search(r'(\d+)\s*(?:месяц|месяца|месяцев)', exp_text)
    if month_match:
        months = int(month_match.group(1))
    
    return years * 12 + months


hh_df['Опыт работы (месяц)'] = hh_df_copy['Опыт работы'].apply(exp_work)

print(f"Максимальный опыт: {hh_df['Опыт работы (месяц)'].max()} месяцев")

print(f"Медиана: {hh_df['Опыт работы (месяц)'].median()} месяцев")
display(hh_df['Опыт работы (месяц)'].describe())
display(hh_df['Опыт работы (месяц)'].head())
#hh_df = hh_df.drop('Опыт работы', axis=1)
display(hh_df.head())
```

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>

```python
#ваш код здесь
busy_categories = ['полная занятость', 'частичная занятость', 'проектная работа', 'волонтерство', 'стажировка']
schedule_categories = ['полный день', 'сменный график', 'гибкий график', 'удалённая работа', 'вахтовый метод']

def sygnal_features(df, column_name, categories_list):
   
    for category in categories_list:
        df[category] = df[column_name].apply(
            lambda x: category in str(x).lower() if pd.notna(x) else False
        )
    return df

hh_df = sygnal_features(hh_df, 'Занятость', busy_categories)
hh_df = sygnal_features(hh_df, 'График', schedule_categories)
employment_columns = [col for col in hh_df.columns if col in busy_categories]
schedule_columns = [col for col in hh_df.columns if col in schedule_categories]
count_p_v = ((hh_df['проектная работа'] == True) & (hh_df['волонтерство'] == True)).sum()
#display(hh_df.head())
count_vahta_flexible = ((hh_df['гибкий график'] == True) & (hh_df['вахтовый метод'] == True)).sum()
print(f"Количество людей, ищущих и проектную работу, и волонтерство: {count_p_v}")
print(f"Количество людей, ищущих и гибкий график, и вахту: {count_vahta_flexible}")
hh_df = hh_df.drop('Занятость', axis=1)
hh_df = hh_df.drop('График', axis=1)

```

4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
#ваш код здесь
cur_df = pd.read_csv('data/ExchangeRates.csv', sep=',')
#display(cur_df.info())
#display(cur_df.isnull().sum())
hh_df['Обновление резюме'] = pd.to_datetime(hh_df['Обновление резюме'], errors='coerce')
hh_df['Дата обновления'] = hh_df['Обновление резюме'].dt.date #не забыть удалить
hh_df['Дата обновления'] = pd.to_datetime(hh_df['Дата обновления'], errors='coerce')
cur_df['date'] = pd.to_datetime(cur_df['date'], errors='coerce')
#display(hh_df['Дата обновления'].head())
#display(cur_df['date'].head())
hh_df[['Сумма_зарплаты', 'Валюта']] = hh_df_copy['ЗП'].str.extract(r'(\d+[\s\d]*) (\D+)')
hh_df['Сумма_зарплаты'] = hh_df['Сумма_зарплаты'].str.replace(' ', '').astype(float)
#display(hh_df.head())
cur_map = {
    'руб.': 'RUB',
    'рублей': 'RUB',
    '₽': 'RUB',
    'руб': 'RUB',
    'тенге': 'KZT',
    'KZT': 'KZT',
    'KGS': 'KGS',
    'сом': 'KGS',
    'сомов': 'KGS',
    'USD': 'USD',
    'долларов': 'USD',
    '$': 'USD',
    'евро': 'EUR',
    'EUR': 'EUR',
    '€': 'EUR',
    'бел.руб.': 'BYR',
    'белорусских рублей': 'BYR',
    'грн.': 'UAH',
    'гривен': 'UAH'
}
hh_df['Валюта_ISO'] = hh_df['Валюта'].map(cur_map)
#display(hh_df[['ЗП', 'Сумма_зарплаты', 'Валюта', 'Валюта_ISO']].head(10))

rub = pd.DataFrame([{'currency': 'RUB', 'close': 1.0, 'date': date} 
                       for date in cur_df['date'].unique()])
cur_df = pd.concat([cur_df, rub], ignore_index=True)
# Проблемный кусок кода:
cur_dict = {}
for _, row in cur_df.iterrows():
    key = (row['date'], row['currency'])
    cur_dict[key] = {'close': row['close'], 'proportion': row['proportion']}

for date in cur_df['date'].unique():
    key = (date, 'RUB')
    cur_dict[key] = {'close': 1.0, 'proportion': 1}

def calculate_rub_salary(row):
    key = (row['Дата обновления'], row['Валюта_ISO'])
    if key in cur_dict:
        close = cur_dict[key]['close']
        proportion = cur_dict[key]['proportion']
        return row['Сумма_зарплаты'] * close / proportion
    return row['Сумма_зарплаты']  # Если курс не найден

hh_df['ЗП (руб)'] = hh_df.apply(calculate_rub_salary, axis=1)
#print(hh_df[['Сумма_зарплаты', 'Валюта_ISO', 'ЗП (руб)']].head())
median_salary = round(hh_df['ЗП (руб)'].median(), 2)
print(f'Медианная зарплата: {median_salary}')
#hh_df = hh_df.drop('ЗП', axis=1)
```
